```python
#LOAD: libraries, parameters, functions

import sys
from pathlib import Path
from IPython.display import Markdown, display

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / 'Quantapp' / '__init__.py').is_file():
        project_root = candidate
        break
else:
    raise RuntimeError('Could not locate the Investment Research project root')
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from Quantapp.data import get_country_macro_config, load_country_macro
from Quantapp.secrets import load_project_env, require_secret
from Quantapp.visualization import build_country_macro_figures

load_project_env()
config = get_country_macro_config('south-korea')
macro = load_country_macro(config, fred_key=require_secret('FRED_API_KEY'))
display(macro.summary)
for warning in macro.warnings:
    display(Markdown(f'> **Data note:** {warning}'))
figures = build_country_macro_figures(config, macro.series, macro.cycle)
for figure in figures.values():
    figure.show()
```
